```python
# Function that returns a personalized greeting
def greet(name):
    return f"Hello, {name}! Welcome!"

# Function that returns sum, difference, product, and quotient
def calculate(a, b):
    sum_val = a + b
    diff_val = a - b
    prod_val = a * b
    quot_val = a / b
    return sum_val, diff_val, prod_val, quot_val


# Call the greeting function
message = greet("Alice")
print(message)

# Call the math function
s, d, p, q = calculate(10, 5)

print("Sum:", s)
print("Difference:", d)
print("Product:", p)
print("Quotient:", q)
```
