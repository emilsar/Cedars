#Exercise 1.1.2: List Methods

```python
# Create a shopping list
shopping = ["milk", "eggs", "bread"]
print(f"Starting list: {shopping}")

# TODO: Complete the following operations:
# 1. Add "butter" to the end
shopping.append("butter")
# 2. Add "cheese" and "yogurt" at once
shopping.extend(["cheese", "yogurt"])
# 3. Insert "flour" at position 1
shopping.insert(1, "flour")
# 4. Remove "milk"
shopping.remove("milk")
# 5. Pop the 3'rd item, and print what item was removed with pop()
item_removed = shopping.pop(2)
print("Item removed: ",item_removed)
# Print final list
print (f"Final list: {shopping}")
# Expected:  ['flour', 'eggs', 'butter', 'cheese', 'yogurt']
```

#Exercise 1.2.2: Using Tuples in Functions

```python
# Write a function that returns multiple statistics
def calculate_statistics(numbers):
    """Return min, max, average, and count."""
    # TODO: Calculate and return all four values as a tuple
    return (min(numbers),max(numbers), sum(numbers)/len(numbers),len(numbers))

print (calculate_statistics.__doc__)
numbers = [10,20,30,40,50]
print (calculate_statistics(numbers))
# Call the function with [10, 20, 30, 40, 50]
# Unpack the results into four variables
# Print them in a readable format
```

#Exercise 1.3.2: Iterating Dictionaries

```python
# Create a dictionary of products and prices
products = {
    "apple": 0.99,
    "banana": 0.59,
    "orange": 1.29
}

# TODO: Complete the following:
# 1. Print each product and its price in a formatted way
print( "Products and prices:")
for product, price in products.items():
    print(f"{product}: ${price:.2f}")
#
# 2. Calculate and print the total if buying one of each
total = sum(products.values())
print(f"Total if buying one of each: ${total:.2f}")
#
# 3. Create a new dictionary with items that cost > $1.00
```
