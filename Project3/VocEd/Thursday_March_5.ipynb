```python
import numpy as np

file_path = '/content/cedars/Project3/imagedata/X/0.npy'

try:
    data = np.load(file_path)
    print(f"Data loaded from {file_path}:")
    print(f"Shape: {data.shape}")
    print(f"Data Type: {data.dtype}")
    # Display first few elements if it's not too large
    if data.size < 100:
        print("Content:")
        print(data)
    elif data.ndim > 1 and data.shape[0] > 0:
        print("First row/slice content:")
        print(data.flatten()[:10]) # Display first 10 elements
    else:
        print("Content too large to display partially. Displaying first 10 elements:")
        print(data.flatten()[:10]) # Display first 10 elements
except Exception as e:
    print(f"Error loading the file: {e}")
```
